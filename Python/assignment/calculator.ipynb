```python
def calculator():
    print("Simple calculator")
    print("Enter an expression like 2 + 3 or 4 * 5")
    expr = input("Expression: ").strip()
    try:
        result = eval(expr, {"__builtins__": None}, {})
        print("Result:", result)
    except Exception as e:
        print("Error:", e)

calculator()
```
